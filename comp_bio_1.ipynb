```python
import matplotlib.pyplot as plt
T = 100
N0 = 15
K = 60
r_values = [1.5, 2.2, 2.6, 2.68, 3.0, 3.4]

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
axes = axes.flatten()

for ax, r in zip(axes, r_values):
    N = N0
    N_values = []
    for i in range(T):
        N = N * (2.718281828 ** (r * (1 - N / K)))
        N_values.append(N)

    ax.plot(range(T), N_values)
    ax.axhline(y=K, color='red', linestyle='--', linewidth=1, label=f'K = {K}')
    ax.set_title(f'r = {r}')
    ax.set_xlabel('Generation')
    ax.set_ylabel('N')
    ax.legend()

fig.suptitle('Ricker Model: N₀ = 15, K = 60', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
T = 500
transient = 400
K = 60
N0 = 15
r_values = []
N_eq_values = []
r = 0.1
while r <= 4.0:
    N = N0
    for i in range(T):
        N = N * (2.718281828 ** (r * (1 - N / K)))
        if i >= transient:
            r_values.append(r)
            N_eq_values.append(N)
    r = round(r + 0.01, 3)

plt.figure(figsize=(10, 6))
plt.plot(r_values, N_eq_values, '.', markersize=0.5, color='steelblue')
plt.axhline(y=K, color='red', linestyle='--', linewidth=1, label=f'K = {K}')
plt.xlabel('Parameter r')
plt.ylabel('N (equilibrium)')
plt.title('Bifurcation Diagram Ricker Model (K = 60, N₀ = 15)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
T = 500
transient = 400
K = 60
N0 = 15
r_values = []
N_eq_values = []
r = 0.1
while r <= 4.0:
    N = N0
    for i in range(T):
        N = max(0, round(N * (2.718281828 ** (r * (1 - N / K)))))
        if i >= transient:
            r_values.append(r)
            N_eq_values.append(N)
    r = round(r + 0.01, 3)

plt.figure(figsize=(10, 6))
plt.plot(r_values, N_eq_values, '.', markersize=0.5, color='steelblue')
plt.axhline(y=K, color='red', linestyle='--', linewidth=1, label=f'K = {K}')
plt.xlabel('Parameter r')
plt.ylabel('N (equilibrium)')
plt.title('Bifurcation Diagram Ricker Model (K = 60, N₀ = 15)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

r1 = 0.5
r2 = 0.6
K1 = 50
K2 = 60
b12 = 0.6
b21 = 0.6


N1 = 10
N2 = 10
T = 100
N1_list = [N1]
N2_list = [N2]
t_list = [0]

for t in range(T):
  dN1 = r1 * N1 * (K1 - N1 - b12 * N2)/ K1
  dN2 = r2 * N2 * (K2 - N2 - b21 * N1)/ K2
  N1 = N1 + dN1
  N2 = N2 + dN2

  N1_list.append(N1)
  N2_list.append(N2)
  t_list.append(t)

plt.figure(figsize=(10, 6))
plt.plot(t_list, N1_list)
plt.plot(t_list, N2_list)

plt.xlabel('Time t')
plt.ylabel('Population size N')
plt.legend('N1', 'N2')

plt.show()



```
